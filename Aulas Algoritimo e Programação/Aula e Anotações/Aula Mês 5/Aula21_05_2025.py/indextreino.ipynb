```python
arquivo = open("numeros.txt", "w")
for linha in range(1, 51):
  arquivo.write("%d\n" %linha)
arquivo.close()
```

```python
arquivo = open("numeros.txt", "r")
for linha in arquivo.readlines():
  print(linha)
arquivo.close()
```

```python
nomeFile = input ("Digite o nome do arquivo: ")
try:
  file = open(nomeFile, "r")
  for linha in file:
    campos = linha.split(';')
    print(campos)
  file.close()
except IOError as erro:
  print("Erro: ", erro)
```

```python
arquivoPar = open("ArquivosTxt/numerosPar.txt", "w")
arquivoImpar = open("ArquivosTxt/numerosImpar.txt", "w")
for linha in range(1, 1001):
  if linha % 2 == 0:
    arquivoPar.write("%d\n" %linha)
  else:
    arquivoImpar.write("%d\n" %linha)
arquivoImpar.close()
arquivoPar.close()

```
